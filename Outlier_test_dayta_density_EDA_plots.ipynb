```python
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
import numpy as np
```

```python
image_path = "/content/SeaLake_1027.jpg"

def load_and_process_image(image_path):
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if img is None:
        raise FileNotFoundError(f"Image not found at {image_path}")
    return img.flatten()
```

```python
try:
    pixel_data = load_and_process_image(image_path)
    plt.figure(figsize=(8, 5))
    sns.kdeplot(pixel_data, fill=True, alpha=0.6)
    plt.title("Density Plot for Image Pixel Intensities: Shuffled image 1")
    plt.xlabel("Pixel Intensity")
    plt.ylabel("Density")
    plt.grid(True)
    plt.show()
except FileNotFoundError as e:
    print(e)
```
